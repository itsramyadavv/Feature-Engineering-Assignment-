### 3 Define Machine Learning. What are the main components in Machine Learning?

Machine Learning is a subset of Artificial Intelligence that enables systems to learn from data, identify patterns, and make decisions with minimal human intervention. It's about teaching computers to learn and improve from experience without being explicitly programmed for every possible scenario.

The main components of Machine Learning typically include:

1.  **Data:** The raw material from which the model learns. This data needs to be relevant, clean, and sufficient.
2.  **Model:** The algorithm or mathematical structure that learns from the data. Examples include linear regression, decision trees, neural networks, etc.
3.  **Objective Function (Loss Function):** A function that measures how well the model is performing. The goal during training is to minimize this function.
4.  **Optimization Algorithm:** The method used to adjust the model's parameters to minimize the objective function. Examples include Gradient Descent, Adam, etc.
5.  **Evaluation Metrics:** Measures used to assess the model's performance on unseen data after training. Examples include accuracy, precision, recall, F1-score, Mean Squared Error, etc.

### 4 How does loss value help in determining whether the model is good or not?

The loss value (or cost value) quantifies the difference between the model's predictions and the actual values in the training data. It's a measure of the model's error.

*   **Lower Loss Value:** Generally indicates that the model's predictions are closer to the actual values, suggesting a better fit to the training data.
*   **Higher Loss Value:** Indicates that the model's predictions are further from the actual values, suggesting a poor fit to the training data.

During training, the goal of the optimization algorithm is to iteratively adjust the model's parameters to minimize this loss function. Monitoring the loss value during training helps track the model's learning progress. However, a low loss on the training data alone doesn't guarantee a good model; it's also crucial to evaluate the model on unseen data (validation or test sets) to check for overfitting.

### 5 What are continuous and categorical variables?

Variables in a dataset can be broadly classified into two main types:

*   **Continuous Variables:** These are variables that can take on any value within a given range. They represent measurements and can have an infinite number of possible values between any two given values.
    *   **Examples:** Height, weight, temperature, age (when measured precisely), time, price.

*   **Categorical Variables:** These are variables that can take on a limited and usually fixed number of possible values, representing categories or groups.
    *   **Examples:** Gender (Male, Female, Other), color (Red, Blue, Green), country of origin, type of animal (Dog, Cat, Bird), educational level (High School, Bachelor's, Master's, PhD).

Categorical variables can be further divided into:
    *   **Nominal Variables:** Categories have no natural order or ranking (e.g., color, gender).
    *   **Ordinal Variables:** Categories have a natural order or ranking (e.g., educational level, customer satisfaction rating like Good, Better, Best).

### 6 How do we handle categorical variables in Machine Learning? What are the common techniques?

Most machine learning algorithms require numerical input. Therefore, categorical variables need to be converted into numerical representations before they can be used to train a model. Common techniques include:

1.  **One-Hot Encoding:** This is one of the most common techniques for nominal categorical variables. It creates a new binary column for each unique category in the variable. A '1' in a column indicates the presence of that category, and '0' indicates its absence.
    *   **Example:** A 'Color' column with values 'Red', 'Blue', 'Green' would be converted into three new columns: 'Color_Red', 'Color_Blue', 'Color_Green'. If the original value was 'Red', the 'Color_Red' column would be 1, and the others would be 0.
    *   **Pros:** Simple to understand and implement, avoids imposing an artificial order.
    *   **Cons:** Can lead to a large number of new features (high dimensionality) if a categorical variable has many unique categories.

2.  **Label Encoding:** This technique assigns a unique integer to each category.
    *   **Example:** A 'Color' column with values 'Red', 'Blue', 'Green' might be encoded as 0, 1, and 2 respectively.
    *   **Pros:** Simple and doesn't increase the number of features.
    *   **Cons:** Can mislead the model into assuming an ordinal relationship between categories where none exists, which can be problematic for nominal variables. Best suited for ordinal variables where the integer mapping reflects the order.

3.  **Ordinal Encoding:** Similar to Label Encoding but specifically used for ordinal variables. The integer assignment reflects the inherent order of the categories.
    *   **Example:** An 'Educational Level' column with values 'High School', 'Bachelor's', 'Master's', 'PhD' could be encoded as 0, 1, 2, 3 respectively, maintaining the order.

4.  **Target Encoding (or Mean Encoding):** This technique replaces each category with the mean of the target variable for that category. It's often used in regression tasks.
    *   **Pros:** Can capture information about the relationship between the category and the target variable.
    *   **Cons:** Can be prone to overfitting, especially if a category has very few instances. Requires careful implementation and validation.

5.  **Binary Encoding:** A combination of hash encoding and label encoding. It converts categories into binary code and then uses the binary digits as features.

The choice of technique depends on the nature of the categorical variable (nominal or ordinal), the number of unique categories, and the specific machine learning algorithm being used.

### 7 What do you mean by training and testing a dataset?

In machine learning, the process of building and evaluating a model involves splitting the available data into typically two or three sets:

*   **Training Set:** This is the largest portion of the dataset and is used to train the machine learning model. The model learns the patterns and relationships within this data by adjusting its internal parameters (weights and biases) based on the objective function. The model "sees" and learns from the data in the training set.

*   **Testing Set:** This is a separate portion of the dataset that the model has *not* seen during training. It is used to evaluate the model's performance on unseen data. The testing set provides an unbiased estimate of how well the model is likely to generalize to new, real-world data. It helps assess the model's ability to make accurate predictions on data it hasn't encountered before.

Using a separate testing set is crucial to detect **overfitting**, which occurs when a model learns the training data too well, including its noise and peculiarities, and performs poorly on new data. If a model performs very well on the training set but poorly on the testing set, it is likely overfitted.

Sometimes, a third set called a **validation set** is also used during the model development process. The validation set is used to tune the model's hyperparameters (settings that are not learned from the data but set before training, like the learning rate or the number of layers in a neural network). This helps in selecting the best model and hyperparameters without touching the final test set, ensuring the test set remains a truly unbiased measure of the final model's performance.

### 9 What is a Test set?

As explained in question 7, a **Test set** is a portion of the dataset that is held back and not used during the training of a machine learning model. Its primary purpose is to provide an unbiased evaluation of the trained model's performance on unseen data.

After the model has been trained on the training set and potentially fine-tuned using a validation set, it is evaluated on the test set. The metrics calculated on the test set (e.g., accuracy, precision, recall, RMSE) give an indication of how well the model is expected to perform on new, real-world data.

It is crucial *not* to use the test set during the training or hyperparameter tuning phases, as this would lead to an overly optimistic estimate of the model's performance and defeat the purpose of having an independent test set.

## 11 Why do we have to perform EDA before fitting a model to the data?

Exploratory Data Analysis (EDA) is a critical step performed *before* fitting a model for several key reasons:

1.  **Understanding the Data:** EDA helps you get a feel for your dataset. You can identify the types of variables (numerical, categorical), their distributions, central tendencies (mean, median), and spread (variance, standard deviation).
2.  **Identifying Patterns and Relationships:** EDA allows you to visually and statistically explore relationships between variables. This includes finding correlations (positive or negative), identifying trends, and understanding how features relate to the target variable. This understanding guides feature selection and engineering.
3.  **Detecting Anomalies and Outliers:** EDA helps spot unusual data points (outliers) or errors that might skew your model's performance. You can then decide whether to remove, transform, or keep these points.
4.  **Handling Missing Values:** EDA reveals the presence and extent of missing data. Understanding where and why data is missing is crucial for deciding on appropriate imputation strategies.
5.  **Assessing Data Quality:** EDA helps identify inconsistencies, errors, or data entry issues that need to be cleaned before modeling.
6.  **Informing Feature Engineering:** By understanding the data through EDA, you can make informed decisions about creating new features or transforming existing ones to improve model performance.
7.  **Selecting Appropriate Models:** The insights gained from EDA can guide the choice of machine learning algorithms. For example, if relationships are highly non-linear, linear models might not be the best fit.
8.  **Identifying Data Skewness and Distribution Issues:** EDA helps understand the distribution of features and the target variable. Some models perform better with normally distributed data, so this step informs the need for transformations.

In essence, EDA is like examining the ingredients before baking a cake. You need to know what you have, its quality, and how it interacts before you start mixing and baking to ensure a good outcome. Skipping EDA can lead to building a model on flawed data, resulting in poor performance and unreliable insights.

### 12 What is correlation?

Correlation is a statistical measure that describes the strength and direction of the linear relationship between two quantitative variables. It tells you how closely two variables are related and whether they tend to increase or decrease together.

The correlation coefficient, often denoted by 'r' (for Pearson correlation), ranges from -1 to +1:

*   **+1:** Perfect positive linear correlation. As one variable increases, the other increases proportionally.
*   **0:** No linear correlation. There is no consistent linear relationship between the two variables.
*   **-1:** Perfect negative linear correlation. As one variable increases, the other decreases proportionally.

Values between -1 and +1 indicate varying degrees of linear relationship (e.g., 0.7 suggests a strong positive correlation, -0.4 suggests a weak negative correlation).

### 13 What does negative correlation mean?

Negative correlation means that as one variable increases, the other variable tends to decrease. The two variables move in opposite directions.

**Example:** The price of a product and the demand for that product often have a negative correlation. As the price increases, the demand tends to decrease.

Another example is the relationship between the number of hours spent exercising and body fat percentage. Generally, as the number of hours spent exercising increases, body fat percentage tends to decrease, indicating a negative correlation.

Only X is needed, since there are no labels.

✅ Example:

from sklearn.cluster import KMeans

model = KMeans(n_clusters=3)
model.fit(X)   # Only features, no y

# Q21 What is feature scaling? How does it help in Machine Learning?

**Feature Scaling** is a data preprocessing technique used to standardize or normalize the range of independent variables or features of data. In simple terms, it means transforming the values of features so that they are on a similar scale.

🔹 Why is Feature Scaling Important?

Many machine learning algorithms are sensitive to the scale of the input features. If features have very different ranges, algorithms that rely on distance calculations or gradient descent can be heavily influenced by features with larger values, while features with smaller values might be ignored or have less impact.

**How it Helps:**

1.  **Improves Performance of Distance-Based Algorithms:** Algorithms like K-Nearest Neighbors (KNN), Support Vector Machines (SVM) with RBF kernel, and K-Means Clustering use distance measures to find similarities between data points. If features are not scaled, features with larger ranges will dominate the distance calculation, leading to inaccurate results. Scaling ensures all features contribute equally to the distance.
2.  **Speeds up Gradient Descent:** Algorithms that use gradient descent (like Linear Regression, Logistic Regression, Neural Networks) converge much faster when features are scaled. Unscaled features can lead to an elongated cost function surface, making it difficult for the optimizer to find the minimum efficiently. Scaling creates a more spherical surface, allowing for quicker convergence.
3.  **Prevents Domination by Large-Value Features:** Without scaling, features with larger magnitudes can disproportionately influence the model's learning process, even if they are not the most important features. Scaling brings all features into a comparable range, giving each feature a more balanced influence.
4.  **Required by Some Algorithms:** Some algorithms, like Principal Component Analysis (PCA) and regularized linear models (Lasso, Ridge), require features to be on a similar scale to work correctly.

🔹 Common Feature Scaling Techniques:

*   **Standardization (Z-score normalization):** Transforms data to have a mean of 0 and a standard deviation of 1. It's less affected by outliers than normalization.
    *   Formula: `(x - mean) / standard deviation`
*   **Normalization (Min-Max scaling):** Scales data to a fixed range, usually between 0 and 1. It's sensitive to outliers.
    *   Formula: `(x - min) / (max - min)`

In summary, feature scaling is essential for many ML algorithms to ensure fair contribution of all features, improve convergence speed, and enhance overall model performance.
